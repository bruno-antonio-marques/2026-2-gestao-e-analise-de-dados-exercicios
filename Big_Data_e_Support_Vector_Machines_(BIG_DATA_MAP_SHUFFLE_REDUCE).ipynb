### AULA - 08: BIG DATA, MAP, REDUCE.

### 01 - IMPORTAR BIBLIOTECAS

```python
from collections import defaultdict
```

### 02 - OBTER DADOS DE ENTRADA

```python
# ==========================================
# DADOS
# ==========================================

documentos = [
    "estou executando um programa python",
    "python é prático para aprender algoritmos",
    "esse big data não é mais um mistério para mim."
]
```

### 03 - EXECUTAR OPERAÇÃO DE MAP ( ENVIO DOS DADOS RECORTADOS AOS NÓS DE PROCESSAMENTO )

```python
# ==========================================
# MAP
# ==========================================

def mapear(texto):

    resultado = []

    palavras = texto.lower().split()

    for palavra in palavras:

        resultado.append(
            (palavra, 1)
        )

    return resultado


mapeados = []

for documento in documentos:

    resultado = mapear(documento)

    mapeados.extend(resultado)


print("MAP")

for item in mapeados:

    print(item)
```

### 04 - EXECUTAR SHUFFLE ( EMBARALHAR ) - ENVIAR TOKENS ( CHAVE + VALOR ) PARA OS NÓS DE PROCESSAMENTO, ORDENAR E PROCESSAR.

```python
# ==========================================
# SHUFFLE
# ==========================================

agrupados = defaultdict(list)

for palavra, valor in mapeados:

    agrupados[palavra].append(valor)


print("\nSHUFFLE")

for palavra, valores in agrupados.items():

    print(
        palavra,
        "->",
        valores
    )
```

### 05 - EXECUTAR REDUCE - PROCESSAR E GERAR O RESULTADO DOS TOKENS ORGANIZADOS E TRAZER O RELATÓRIO DOS DADOS PROCESSADOS PARA A MÁQUINA DO OPERADOR.

```python
# ==========================================
# REDUCE
# ==========================================

def reduzir(palavra, valores):

    return (
        palavra,
        sum(valores)
    )


resultado_final = []

for palavra, valores in agrupados.items():

    resultado = reduzir(
        palavra,
        valores
    )

    resultado_final.append(resultado)


print("\nREDUCE")

for palavra, quantidade in resultado_final:

    print(
        palavra,
        "->",
        quantidade
    )
```

```python
from collections import defaultdict


# ==========================================
# DADOS
# ==========================================

documentos = [
    "big data usa dados testes",
    "dados geram informacao",
    "big data usa processamento"
]


# ==========================================
# MAP
# ==========================================

def mapear(texto):

    resultado = []

    palavras = texto.lower().split()

    for palavra in palavras:

        resultado.append(
            (palavra, 1)
        )

    return resultado


mapeados = []

for documento in documentos:

    resultado = mapear(documento)

    mapeados.extend(resultado)


print("MAP")

for item in mapeados:

    print(item)


# ==========================================
# SHUFFLE
# ==========================================

agrupados = defaultdict(list)

for palavra, valor in mapeados:

    agrupados[palavra].append(valor)


print("\nSHUFFLE")

for palavra, valores in agrupados.items():

    print(
        palavra,
        "->",
        valores
    )


# ==========================================
# REDUCE
# ==========================================

def reduzir(palavra, valores):

    return (
        palavra,
        sum(valores)
    )


resultado_final = []

for palavra, valores in agrupados.items():

    resultado = reduzir(
        palavra,
        valores
    )

    resultado_final.append(resultado)


print("\nREDUCE")

for palavra, quantidade in resultado_final:

    print(
        palavra,
        "->",
        quantidade
    )
```
